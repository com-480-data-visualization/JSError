```python
def merge_save(base):
    df1 = pd.read_csv(f'{base}_east.csv')
    df2 = pd.read_csv(f'{base}_west.csv')

    df = pd.concat([df1, df2], axis=0, ignore_index=True)
    print(df.head())
    df.to_csv(f'{base}.csv', index=False)
```

```python
label_mapping = {}

if meta_west.variable_value_labels.get('Q1'):
    q1_mapping = meta_west.variable_value_labels.get('Q1')
    # Create a new column with the readable labels
    # df['Q1_readable'] = df['Q1'].map(q1_mapping)

if meta_west.variable_value_labels.get('QCURRELrec'):
    label_mapping = meta_west.variable_value_labels.get('QCURRELrec')
    
print("Label Mapping:", label_mapping)
if not label_mapping:
    print("No value labels found for QCURRELrec. Using raw codes.")

# 2. Determine the order of religions
df_mapped = df_west.copy()

df_mapped['QCURRELrec'] = df_west['QCURRELrec'].map(label_mapping).fillna(df_west['QCURRELrec'])

#    Option B: keep both, adding a new "readable" column
# df['religion_readable'] = df['QCURRELrec'].map(label_mapping).fillna(df['QCURRELrec'])

# 3. (Optional) If you want to rename the column for clarity
df_renamed = df_mapped.rename(columns={'QCURRELrec': 'religion'})
df_renamed = df_renamed.drop(columns=['country'])
df_renamed = df_renamed.rename(columns={'COUNTRY_readable': 'country'})

mask = df_renamed['religion'].str.contains('Refused', na=False)
df_renamed = df_renamed[~mask].copy()

df_renamed['religion'] = df_renamed['religion'].replace({
    'Muslim+Jewish+Buddhist+Hindu+something else': 'Other',
})

# 3. Build counts by country × religion
df_counts = (
    df_renamed
    .groupby(['country', 'religion'])
    .size()
    .reset_index(name='count')
)

# 4. Compute within-country percentages
df_counts['percentage'] = (
    df_counts
      .groupby('country')['count']
      .transform(lambda x: x / x.sum() * 100)
      .round(2)
)

df_counts = df_counts.drop(columns='count')



df_counts.to_csv('religions_cleaned_west.csv', index=False)
```

##### East

```python
religion_cols = [col for col in df_east.columns if col.startswith('qcurrel') and col.endswith('rec')]

df_list = []

# Loop over each religion column, replace numeric codes with text labels and create a country indicator
for col in religion_cols:
    # Attempt to get the mapping for this column
    mapping = meta_east.variable_value_labels.get(col)
    if mapping is None:
        print(f"No value labels found for {col}. Using original values.")
        mapped_series = df_east[col]
    else:
        mapped_series = df_east[col].map(mapping)
    
    # Create a temporary DataFrame.
    # Optionally, extract the country code from the column name.
    # For example, if col is 'qcurrelromrec', you might extract 'rom' as:
    country = col.replace('qcurrel', '').replace('rec', '')
    
    temp_df = pd.DataFrame({
        'religion': mapped_series,
        'country': country
    })
    df_list.append(temp_df)

# Combine all the temporary DataFrames into one long DataFrame
df_combined = pd.concat(df_list, ignore_index=True)

# Optional: Remove missing values if needed
df_combined = df_combined.dropna(subset=['religion'])

mask = df_combined['religion'].str.contains('DO NOT READ', na=False)
df_clean = df_combined[~mask].copy()

# 2 & 3. Replace the various labels
df_clean['religion'] = df_clean['religion'].replace({
    # combine misspelled and correct versions into "Other"
    'Soemthing else/other': 'Other',
    'Something else/other': 'Other',
    # unify the "ancient" label under the old believer category
    'Ancient religion or traditional religion': 'Old believer/old ritualist'
})


df_grouped = (
    df_clean
    .groupby(['country', 'religion'])
    .size()                       # compute counts
    .reset_index(name='count')    # turn the Series into a DataFrame
)

df_grouped['percentage'] = (
    df_grouped['count']
      .groupby(df_grouped['country'])
      .transform(lambda x: x / x.sum() * 100)
      .round(2)   # rounds to 2 decimal places
)

# 2. (Optional) If you only want country/religion/percentage:
df_pct = df_grouped.drop(columns='count')


print(df_pct)

df_pct.to_csv('religions_cleaned_east.csv', index=False)
```

### Economic state perception


```python
mapping_q2 = meta_east.variable_value_labels.get('Q2')
if mapping_q2 is None:
    print("No value labels found for Q2. Using original numeric values.")
    df_east['Q2_text'] = df_east['Q2']
else:
    df_east['Q2_text'] = df_east['Q2'].map(mapping_q2)

# Check unique responses after mapping
# print("Unique economic state responses:", df['Q2_text'].dropna().unique())

# 3. drop any answers with "DO NOT READ"
df_clean = df_east[~df_east['Q2_text'].str.contains("DO NOT READ", na=False)]

# 4. group & count
grp = (
    df_clean
    .groupby(['COUNTRY_readable', 'Q2_text'])
    .size()
    .reset_index(name='count')
)

# 5. compute percentage within each country
grp['percentage'] = (
    grp['count']
    / grp.groupby('COUNTRY_readable')['count'].transform('sum')
    * 100
)

# 6. rename columns for clarity
grp = grp.rename(columns={
    'COUNTRY_readable': 'country',
    'Q2_text':    'economic_perception'
})

# 7. keep only the three columns you want
out = grp[['country', 'economic_perception', 'percentage']]

# 8. save to CSV
out.to_csv('economic_state_percentages_east.csv', index=False)

print("Saved percentages by country to economic_state_percentages.csv")
```
